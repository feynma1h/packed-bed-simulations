```python
x = [];
N = 10
for i in range(-N, N+1):
    for j in range(-N, N+1):
        x.append((i, j));

import numpy as np
import math
R = np.zeros(N+1) # stores the number of pixels cut by circles having radius 1, 2, ... N
for i in range(-N, N+1):
    for j in range(-N, N+1):
        if i==0:
            xmin=0; xmax=0.5;
        else:
            xmin=min(abs(i+0.5), abs(i-0.5)); xmax=max(abs(i+0.5), abs(i-0.5));
        if j==0:
            ymin=0; ymax=0.5;
        else:
            ymin=min(abs(j+0.5), abs(j-0.5)); ymax=max(abs(j+0.5), abs(j-0.5));
        for r in range(1, N+1):
            if math.hypot(xmin, ymin)<r<math.hypot(xmax, ymax):
                R[r]+=1
```
